```python

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report)
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
model = Sequential([
    Flatten(input_shape=(28,28)),

    Dense(128, activation='relu'),

    Dense(64, activation='relu'),

    Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(X_train,y_train,epochs=5,validation_split=0.2)
```

```python
loss, accuracy = model.evaluate(X_test,y_test)
print("\nTest Accuracy:", round(accuracy*100,2), "%")
```

```python
X_small, _, y_small, _ = train_test_split(X_test,y_test,test_size=0.90,random_state=42,stratify=y_test)

print("\n10% Test Data Shape:")
print(X_small.shape)
```

```python
y_prob = model.predict(X_small)

y_pred = np.argmax(y_prob,axis=1)
```

```python
acc = accuracy_score(y_small,y_pred)

prec = precision_score(y_small,y_pred,average='weighted')

rec = recall_score(y_small,y_pred,average='weighted')

f1 = f1_score(y_small,y_pred,average='weighted')
print("\nAccuracy :", round(acc*100,2), "%")
print("Precision:", round(prec*100,2), "%")
print("Recall   :", round(rec*100,2), "%")
print("F1 Score :", round(f1*100,2), "%")
```

```python
cm = confusion_matrix(y_small,y_pred)
print("\nConfusion Matrix:")
print(cm)
```

```python
print("\nClassification Report:\n")

print(classification_report(y_small,y_pred))
```
